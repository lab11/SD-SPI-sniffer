```python
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("classic")

# Load the CSV file into a DataFrame
file_path = "logic_analyzer_trace.csv"  # Replace with your file path
data = pd.read_csv(file_path)


# Filter the data between 1.7s and 1.8s
filtered_data = data[(data['Time [s]'] >= 1.663) & (data['Time [s]'] <= 1.6645)]

# Check if the filtered data is empty
if filtered_data.empty:
    print("No data found in the specified time range (1.7s - 1.8s).")
else:
    # Create a figure with 4 subplots
    fig, axs = plt.subplots(4, 1, figsize=(15, 6), sharex=True)

    # Plot each signal in its respective subplot
    axs[0].step(filtered_data['Time [s]'], filtered_data['CS'], where='post',color='b')
    axs[0].set_title('CS Signal')
    axs[0].set_ylabel('CS')
    axs[0].set_ylim(-0.05,1.05)

    axs[1].step(filtered_data['Time [s]'], filtered_data['CLK'], color='g')
    axs[1].set_title('CLK Signal')
    axs[1].set_ylabel('CLK')
    axs[1].set_ylim(-0.05,1.05)

    axs[2].step(filtered_data['Time [s]'], filtered_data['MOSI'], color='r')
    axs[2].set_title('MOSI Signal')
    axs[2].set_ylabel('MOSI')
    axs[2].set_ylim(-0.05,1.05)

    axs[3].step(filtered_data['Time [s]'], filtered_data['MISO'], color='m')
    axs[3].set_title('MISO Signal')
    axs[3].set_ylabel('MISO')
    axs[3].set_xlabel('Time [s]')
    axs[3].set_ylim(-0.05,1.05)

    # Adjust layout for clarity
    plt.tight_layout()
    plt.show()
```
